```python
def count_vowels(text):
  counts={
      "a":0,
      "e":0,
      "i":0,
      "o":0,
      "u":0,
  }
  text= text.lower()

  for character in text:
    if character in counts:
      counts[character] +=1

  return counts

word = input()
print(count_vowels(word))
```

```python
def is_prime(number):
  if number < 2:
      return False

  for i in range(2,number):
    if number % i == 0:
      return False

  return True

num = int(input())
is_prime(num)
```

```python
def is_palindrome(text):
    text = text.lower()

    if text == text[::-1]:
        return True
    else:
        return False
word = input()
print(is_palindrome(word))
def is_symmetrical(text):
    mid = len(text) // 2

    if text[:mid] == text[mid:]:
        return True
    else:
        return False
word = input()
print(is_symmetrical(word))
```

```python
def sum_list(numbers):
  total = 0

  for number in numbers:
    total += number

  return total

numbers = list(map(int, input().split()))
print(sum_list(numbers))
```

```python
class operations:
    def __init__(self, number):
        self.number = number

    def addition(self, number):
        return self.number + number

    def subtraction(self, number):
        return self.number - number

    def multiplication(self, number):
        return self.number * number

    def division(self, number):
        if number == 0:
            return "Cannot divide by zero"
        return self.number / number
number1 = int(input())
number2 = int(input())

obj = operations(number1)

print(obj.addition(number2))
print(obj.subtraction(number2))
print(obj.multiplication(number2))
print(obj.division(number2))
```

```python
def separate_even_odd(numbers):
    even = []
    odd = []

    for number in numbers:
        if number % 2 == 0:
            even.append(number)
        else:
            odd.append(number)

    return even, odd

nums = [int(i) for i in input().split(',')]
print(separate_even_odd(nums))
```

```python
def maximum_marks(marks):
    max_mark = 0
    max_subject = ""

    for subject, mark in marks.items():
        if mark > max_mark:
            max_mark = mark
            max_subject = subject

    return max_subject
marks = {
    "math": int(input()),
    "english": int(input()),
    "science": int(input()),
    "computer": int(input())
}

print(maximum_marks(marks))
```

```python
def is_anagram(text1, text2):
    text1 = text1.lower()
    text2 = text2.lower()

    return sorted(text1) == sorted(text2)

a = input()
b = input()
print(is_anagram(a,b))
```

```python
def max_profit(prices):
    min_price = prices[0]
    max_profit = 0

    for price in prices:
        if price < min_price:
            min_price = price

        profit = price - min_price

        if profit > max_profit:
            max_profit = profit

    return max_profit
prices = list(map(int, input().split()))

print(max_profit(prices))
```

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        if num - 1 not in num_set:
            current = num
            length = 1

            while current + 1 in num_set:
                current += 1
                length += 1

            longest = max(longest, length)

    return longest
nums = list(map(int, input().split()))

print(longest_consecutive(nums))
```
